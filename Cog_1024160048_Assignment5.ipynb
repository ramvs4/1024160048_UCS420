```python
# Q4
ram = np.linspace(10, 100, 25)
print(ram)
print("Dimensions:", ram.ndim)
print("Shape:", ram.shape)
print("Total elements:", ram.size)
print("Data type:",ram.dtype)
print("Total bytes:",ram.nbytes)

transpose = ram.reshape(25, 1)
print("Transpose :")
print(transpose)

print("Transpose using T:")
print(ram.T)
```
