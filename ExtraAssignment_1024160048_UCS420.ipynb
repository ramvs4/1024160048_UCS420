```python
def pole(Pole_Positions):
  if Pole_Positions>5:
    return "Chmapion"
  elif Pole_Positions>=4:
    return "Winner"
  else :
    return "Racer"
df["Review"]=df['Pole_Positions'].apply(pole)
print(df.head())
```
