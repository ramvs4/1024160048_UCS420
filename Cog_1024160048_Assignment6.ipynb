```python
# Q1
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
temperature=temperature+2
print(temperature)
new_temp=9*(temperature/5)+32
print(new_temp)
t=temperature[temperature>32]
print(t)
print(np.sum(temperature>32))
```

```python
# Q2
import numpy as np
steps = np.array([[5000, 6200, 7100], [8000, 7500, 9000], [4500, 5100, 4800], [9000, 8500, 9500]])
print(steps.sum())
print(np.mean(steps))
print(np.max(steps))
print(np.min(steps))
print(np.sum(steps,axis=0))
print(np.sum(steps,axis=1))
np.unravel_index(np.argmax(steps),steps.shape)
```

```python
# Q4
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])
print(X.shape)
print(X.ndim)
print(X.T)
XtX = X.T @ X
print(XtX)
inverse = np.linalg.inv(XtX)
print(inverse)
beta = inverse @ X.T @ y
print(beta)
new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print(prediction)
```
